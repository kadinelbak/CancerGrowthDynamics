# Optimal Control One

Compare simple growth, delayed-loss, and Gevertz et al. (2025) drug-induced-resistance candidates with one joint BIC across the A2780 30k treated-monoculture data. The eligible BIC winner is then validated at the held-out 1.0 uM dose and used for equal-budget in vitro schedule optimization. Latent resistant cells, when present, are model states rather than measured A2780cis cells.

```julia
using MechanicalAutomaticModeling
using CSV, DataFrames
```

The processed data retain historical endpoint folder names. The package correction is `legacy IC25 -> 1.47 uM`, `IC50 -> 1.0 uM`, and `legacy IC75 -> 0.67 uM`.

```julia
package_root = normpath(joinpath(@__DIR__, ".."))
result = MechanicalAutomaticModeling.OptimalControlOne.run_optimal_control_one!(
    start = package_root,
    fit_seconds = 8.0,
    control_seconds = 20.0,
)
```

```julia
fits = CSV.read(joinpath(result.output, "dose_specific_fits.csv"), DataFrame)
control = CSV.read(joinpath(result.output, "control_metrics.csv"), DataFrame)
fits, control, result.report
```

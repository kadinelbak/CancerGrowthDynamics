# Optimal Control One

Audit the complete four-stage A2780 inheritance chain before attempting optimal control: untreated monoculture, treated monoculture, untreated coculture, and treated coculture. Each stage reports BIC and trajectory error for competing models. Treated-coculture candidates are additionally checked against day-14 bootstrap intervals. Optimal control is generated only after the inherited chain passes the validation gate.

```julia
using MechanicalAutomaticModeling
using CSV, DataFrames
```

The previous endpoint-only interpolation at 1.0 uM is excluded from validation because the measured 1.0 uM trajectory is not intermediate between 0.67 and 1.47 uM. The staged exports remain the source of model rankings, inherited parameters, fitted overlays, and bootstrap intervals.

```julia
package_root = normpath(joinpath(@__DIR__, ".."))
result = MechanicalAutomaticModeling.OptimalControlOne.run_optimal_control_one!(
    start = package_root,
)
```

```julia
stage_models = CSV.read(joinpath(result.output, "four_stage_model_audit.csv"), DataFrame)
stage4_endpoints = CSV.read(joinpath(result.output, "stage4_endpoint_model_audit.csv"), DataFrame)
readiness = CSV.read(joinpath(result.output, "control_readiness.csv"), DataFrame)
stage_models, stage4_endpoints, readiness, result.report
```

## **Shared Functions**

```math
g(N) = r\left(1 - \left(\frac{N}{K}\right)^v\right)
```

```math
H(d) = \frac{d^n}{IC50^n + d^n}
```

```math
G(t) =
\begin{cases}
0, & t \le t_{\text{onset}} \\[6pt]
1 - e^{-\lambda (t - t_{\text{onset}})}, & t > t_{\text{onset}}
\end{cases}
```

Special case when the model does not include (t_{\text{onset}}):

```math
G(t) = 1 - e^{-\lambda t}
```

# **Hill_ramp**

```math
E(t) = E_{\max}\,H(d)\,G(t)
```

```math
\frac{dN}{dt}
= g(N)\,\bigl(1 - E(t)\bigr)
```

```math
N(0) = N_0
```

# **Model1_parpi**

```math
N(t) = P(t) + A(t)
```

```math
\frac{dP}{dt}
= g(N)\,(1 - 2\alpha)\,P + \beta\,A
```

```math
\frac{dA}{dt}
= \alpha\,g(N)\,P - \beta\,A - d\,A
```

```math
P(0) = N_0
```

```math
A(0) = 0
```

# **Model2_parpi**

```math
N(t) = \sum_{i=0}^{n} P_i(t) + A(t)
```

For (P_0):

```math
\frac{dP_0}{dt}
= g(N)\,(1 - 2\alpha)\,P_0 + \beta\,P_1
```

For intermediate compartments (P_i):

```math
\frac{dP_i}{dt}
= 2\alpha\,g(N)\,P_{i-1}
+ g(N)\,(1 - 2\alpha)\,P_i
- \beta\,P_i
+ \beta\,P_{i+1}
```

For final compartment (P_n):

```math
\frac{dP_n}{dt}
= 2\alpha\,g(N)\,P_{n-1}
+ g(N)\,(1 - 2\alpha)\,P_n
- \beta\,P_n
+ \beta\,A
```

For apoptotic (A):

```math
\frac{dA}{dt}
= \alpha\,g(N)\,P_n - \beta\,A - d\,A
```

Initial:

```math
P_0(0) = N_0
```

```math
P_i(0) = 0\quad (i \ge 1)
```

```math
A(0) = 0
```

# **two_pop_ramp_tonset_switch**

```math
N(t) = N_s(t) + N_r(t)
```

```math
E_s(t) = E_{\text{sens}}\,H(d)\,G(t)
```

```math
E_r(t) = E_{\text{res}}\,H(d)\,G(t)
```

```math
S_{\rightarrow R}(t) = q_{sr}\,G(t)\,N_s
```

```math
\frac{dN_s}{dt}
= g(N)\,(1 - E_s(t))\,N_s
- \delta_s\,G(t)\,N_s
- S_{\rightarrow R}(t)
```

```math
\frac{dN_r}{dt}
= g(N)\,(1 - E_r(t))\,N_r
- \delta_r\,G(t)\,N_r
+ S_{\rightarrow R}(t)
```

```math
N_s(0) = f_{s0}\,N_0
```

```math
N_r(0) = (1 - f_{s0})\,N_0
```

## **Co-Culture Model Equations (All Models)**

### **Model 1 — Full Theta-Logistic LV (separate θ and death)**  
\[
\begin{aligned}
\frac{dS}{dt} &= r_S S \left(1 - \left(\frac{S+R}{K}\right)^{v_S}\right) - d_S S, \\[6pt]
\frac{dR}{dt} &= r_R R \left(1 - \left(\frac{S+R}{K}\right)^{v_R}\right) - d_R R.
\end{aligned}
\]

### **Model 2 — Theta-Logistic LV (shared θ, no death)**  
\[
\begin{aligned}
N &= S + R, \\[6pt]
\frac{dS}{dt} &= r_S S \left(1 - \left(\frac{N}{K}\right)^{v}\right), \\[6pt]
\frac{dR}{dt} &= r_R R \left(1 - \left(\frac{N}{K}\right)^{v}\right).
\end{aligned}
\]

### **Model 3 — Logistic Competition (shared K, no θ)**
\[
\begin{aligned}
N &= S + R, \\[6pt]
\frac{dS}{dt} &= r_S S \left(1 - \frac{N}{K}\right), \\[6pt]
\frac{dR}{dt} &= r_R R \left(1 - \frac{N}{K}\right).
\end{aligned}
\]

### **Model 4 — Generalized LV (symmetric competition)**
\[
\begin{aligned}
\frac{dS}{dt} &= r_S S \left(1 - \frac{S + \alpha R}{K}\right), \\[6pt]
\frac{dR}{dt} &= r_R R \left(1 - \frac{\alpha S + R}{K}\right).
\end{aligned}
\]

### **Model 5 — Generalized LV (symmetric competition + death)**
\[
\begin{aligned}
\frac{dS}{dt} &= r_S S \left(1 - \frac{S + \alpha R}{K}\right) - d_S S, \\[6pt]
\frac{dR}{dt} &= r_R R \left(1 - \frac{\alpha S + R}{K}\right) - d_R R.
\end{aligned}
\]
